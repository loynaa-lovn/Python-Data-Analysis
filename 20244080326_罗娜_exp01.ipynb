```python
#1 score_level
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0‑100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(88))
print(score_level(70))
print(score_level(50))

#2 for、while求1‑100和
s_for = 0
for i in range(1,101):
    s_for += i
print("for求和1‑100", s_for)

s_while = 0
i=1
while i<=100:
    s_while += i
    i +=1
print("while求和1‑100", s_while)

#3 阶乘：迭代、递归
def fact_iter(n):
    res = 1
    for k in range(2, n+1):
        res *=k
    return res

def fact_rec(n):
    if n ==0 or n ==1:
        return 1
    return n * fact_rec(n-1)

for n in [0,1,5,10]:
    print(f"n={n}, iter={fact_iter(n)}, rec={fact_rec(n)}")

#4 summary 返回最小、最大、平均
def summary(values):
    minv = min(values)
    maxv = max(values)
    avg = sum(values)/len(values)
    return (minv, maxv, avg)

res = summary([10,20,30])
mi,ma,av = res
print("summary解包", mi, ma, av)

#5 sum_all可变参数
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1,2,3,4))
```

实验 3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

#1 zip构建字典，items输出
stu_dict = dict(zip(names, scores))
for k,v in stu_dict.items():
    print(k, v)
print("      ")

#2 enumerate序号从1开始
for idx, (n,s) in enumerate(zip(names, scores), start=1):
    print(idx, n, s)
print("      ")


#3 列表推导式
pass_names = [n for n,s in zip(names,scores) if s>=60]
score_sq = [s**2 for s in scores]
print("及格姓名", pass_names)
print("成绩平方", score_sq)
print("      ")


#4 统计
cnt = len(scores)
avg = sum(scores)/cnt
high = max(scores)
low = min(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / cnt
print(f"人数:{cnt},平均分:{avg:.2f},最高分:{high},最低分:{low},及格率:{pass_rate:.2%}")
print("      ")


#5 对比 alias vs copy
demo = [10,20,30]
alias = demo
copied = demo.copy()
alias[0] = 99
print("demo", demo)
print("alias", alias)
print("copied", copied)
print("      ")



#6 元组不可修改；zip长短不一
t = ("Tom", 88)
try:
    t[1]=99
except TypeError as e:
    print("元组赋值异常",e)

list1 = [1,2]
list2 = [10,20,30]
print(list(zip(list1,list2))) # 以短序列为准
```

实验4：⽂件读写、异常处理与模块

```python
import os

# 任务1：查看工作目录，构造路径，创建数据目录 
cwd = os.getcwd()
print("当前工作目录：", cwd)

data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据文件夹路径：", data_dir)

print()

#任务2：with open写入students.txt，读取文件，跳过表头 
txt_path = os.path.join(data_dir, "students.txt")

content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("读取到全部行：")
for line in lines:
    print(repr(line))

print()

#  任务3：编写 parse_student(line) 函数 
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数：{score_str}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出0~100范围：{score}")
    return {"name": name, "score": score}

print("parse_student函数定义完成")
print()

# 任务4：逐行处理，捕获异常，收集有效记录
valid_records = []
error_log = []

for lineno, line in enumerate(lines, start=1):
    if lineno == 1:
        continue
    try:
        res = parse_student(line)
        valid_records.append(res)
    except ValueError as e:
        error_log.append((lineno, str(e)))

print(" 有效记录：")
for item in valid_records:
    print(item)

print("\n 错误记录：")
for no, msg in error_log:
    print(f"行{no}：{msg}")

print()

import os
import pickle

# 任务5：写入report.txt，演示 w 和 a 模式
data_dir = os.path.join(os.getcwd(), "exp01_data")
report_path = os.path.join(data_dir, "report.txt")

valid_num = len(valid_records)
invalid_num = len(error_log)

if valid_num > 0:
    avg_score = sum(i["score"] for i in valid_records) / valid_num
    pass_cnt = sum(1 for i in valid_records if i["score"] >= 60)
    pass_rate = pass_cnt / valid_num
else:
    avg_score = 0.0
    pass_rate = 0.0

# w模式：覆盖
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录:{valid_num}\n")
    f.write(f"无效记录:{invalid_num}\n")
    f.write(f"平均分:{avg_score:.2f}\n")
    f.write(f"及格率:{pass_rate:.2%}\n")
print("w模式写入完成")

# a模式：追加
with open(report_path, "a", encoding="utf-8") as f:
    f.write("==== a模式追加内容 ====\n")
print("a模式追加完成")

print()

#任务6：pickle保存与读取，校验相等 
pkl_path = os.path.join(data_dir, "students.pkl")

with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)

print("pickle读取后和原数据是否相等：", loaded == valid_records)

print()

# 任务7：从 student_utils.py 导入函数测试 
from student_utils import parse_student

print("有效输入测试：", parse_student("Tom,77"))
try:
    parse_student("Lily,200")
except ValueError as e:
    print("无效输入测试异常：", e)

print()

# 任务8：读取不存在文件，捕获FileNotFoundError 
not_exist_file = os.path.join(data_dir, "no_such_file.txt")
try:
    with open(not_exist_file, "r", encoding="utf-8") as f:
        pass
except FileNotFoundError as e:
    print("捕获FileNotFoundError：", e)
```
