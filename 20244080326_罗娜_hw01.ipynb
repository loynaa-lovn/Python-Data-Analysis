```python
#欧几里得算法GCD
def gcd(m,n):
    """欧几里得算法求最大公约数"""
    while n != 0:
        m, n = n, m % n
    return m

# 测试
print("gcd(2018,2019) =", gcd(2018,2019))
print("gcd(1200,300) =", gcd(1200,300))
print("gcd(5040,60) =", gcd(5040,60))

# 负数测试
print("\n负数测试：")
print("gcd(-12, 18) =", gcd(-12,18))
print("gcd(-24,-18) =", gcd(-24,-18))
```

```python
#近似欧拉数e
import math
# 1.伯努利极限 e ≈ (1+1/n)^n
def euler_limit(n):
    return (1 + 1.0 / n) ** n

# 2.级数前n项和 e=sum_{k=0}^{n-1} 1/k!
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1   # 0! =1
    for k in range(n):
        if k ==0:
            fact =1
        else:
            fact *= k
        total += 1.0 / fact
    return total

# 3. euler_approx：误差小于ϵ，while循环
def euler_approx(epsilon):
    true_e = math.exp(1)
    n = 1
    while True:
        approx = euler_infinite_sum(n)
        if abs(approx - true_e) < epsilon:
            return approx
        n +=1

#4.打印表格，k从1到n输出euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")


#测试
print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4)) #≈2.66666
print("euler_approx(0.001):", euler_approx(0.001))
print("\n---print_euler_sum_table(6)---")
print_euler_sum_table(6)
print("math.exp(1)真实e：", math.exp(1))
```

```python
#测试整数属性
#1 判断n是否2的幂
def is_power_of_2(n):
    while n % 2 == 0:
        n = n //2
    return n == 1

#2 n是否b的幂
def is_power(b, n):
    while n % b == 0:
        n = n // b
    return n ==1

#测试
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(7):", is_power_of_2(7))
print("is_power(3,81):", is_power(3,81))
print("is_power(5,100):", is_power(5,100))
```
